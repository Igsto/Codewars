# Decimal to any Rational or Irrational Base Converter

```python
from math import pi, log, floor, ceil

def converter(n, decimals=0, base=pi):
    chr_list = {i: str(i) if i < 10 else chr(i+55) for i in range(ceil(base))}
    n1 = abs(n)
    if n == 0:
        return '0.' + '0' * decimals
    k = floor(log(n1, base)) + 1
    b = floor(base)
    if n > 0:
        base_int = ''
    else:
        base_int = '-'
    if k < 1:
        base_int += '0'
    

    for l in range(k-1,(decimals+1)* -1, -1):
        if l == -1:
            base_int += '.'
        ai = floor(n1/base**l)
        base_int += chr_list[ai]
        n1 -= ai * base**l
        
    return base_int
```

```python
converter(pi)
```
